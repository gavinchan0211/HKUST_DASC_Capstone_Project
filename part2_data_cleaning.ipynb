```python
import pandas as pd
import numpy as np

# Load the dataset
file_path = "sleep_cycle_productivity.csv"
df = pd.read_csv(file_path)

# Step 1: Convert 'Gender' to numbers
gender_mapping = {"Male": 1, "Female": 2, "Other": 3}
df["Gender"] = df["Gender"].map(gender_mapping)

# Step 2: Add standardized columns next to their corresponding original columns
# Create a function to insert a column after another
def insert_column_after(df, new_column_name, new_column_data, after_column_name):
    col_position = df.columns.get_loc(after_column_name) + 1
    df.insert(col_position, new_column_name, new_column_data)

# Standardize 'Total Sleep Hours' and insert it after 'Total Sleep Hours'
standardized_sleep_hours = ((df["Total Sleep Hours"] - df["Total Sleep Hours"].min()) /
                            (df["Total Sleep Hours"].max() - df["Total Sleep Hours"].min()) * 100).round(5)
insert_column_after(df, "Standardized Total Sleep Hours", standardized_sleep_hours, "Total Sleep Hours")

# Standardize 'Exercise (mins/day)' and insert it after 'Exercise (mins/day)'
standardized_exercise = ((df["Exercise (mins/day)"] - df["Exercise (mins/day)"].min()) /
                         (df["Exercise (mins/day)"].max() - df["Exercise (mins/day)"].min()) * 100).round(5)
insert_column_after(df, "Standardized Exercise", standardized_exercise, "Exercise (mins/day)")

# Standardize 'Caffeine Intake (mg)' and insert it after 'Caffeine Intake (mg)'
standardized_caffeine = ((df["Caffeine Intake (mg)"] - df["Caffeine Intake (mg)"].min()) /
                         (df["Caffeine Intake (mg)"].max() - df["Caffeine Intake (mg)"].min()) * 100).round(5)
insert_column_after(df, "Standardized Caffeine Intake", standardized_caffeine, "Caffeine Intake (mg)")

# Standardize 'Screen Time Before Bed (mins)' and insert it after 'Screen Time Before Bed (mins)'
standardized_screen_time = ((df["Screen Time Before Bed (mins)"] - df["Screen Time Before Bed (mins)"].min()) /
                            (df["Screen Time Before Bed (mins)"].max() - df["Screen Time Before Bed (mins)"].min()) * 100).round(5)
insert_column_after(df, "Standardized Screen Time Before Bed", standardized_screen_time, "Screen Time Before Bed (mins)")

# Standardize 'Work Hours (hrs/day)' and insert it after 'Work Hours (hrs/day)'
standardized_work_hours = ((df["Work Hours (hrs/day)"] - df["Work Hours (hrs/day)"].min()) /
                           (df["Work Hours (hrs/day)"].max() - df["Work Hours (hrs/day)"].min()) * 100).round(5)
insert_column_after(df, "Standardized Work Hours", standardized_work_hours, "Work Hours (hrs/day)")

# Save the cleaned and modified dataset
cleaned_file_path = "sleep_cycle_productivity_cleaned.csv"
df.to_csv(cleaned_file_path, index=False)

print(f"Cleaned and modified data saved to {cleaned_file_path}")
```
